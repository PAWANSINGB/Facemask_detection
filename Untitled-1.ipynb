```python
import tensorflow as tf
from tensorflow.keras import layers, models
```

```python
# 1. Path & Parameters 
DATA_DIR = r"C:\Users\pawan singh\OneDrive\Desktop\Face Mask Detection\archive\data"
IMAGE_SIZE = (128, 128)
BATCH_SIZE = 32
EPOCHS = 10
```

```python
# 2. Entire Dataset Load 
raw_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR,
    seed=123,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE
)
```

```python
# Dataset ki Total Batches Count
dataset_size = tf.data.experimental.cardinality(raw_ds).numpy()
```

```python
# 70% Train, 15% Validation, 15% Test Size Calculate Karna
train_size = int(0.7 * dataset_size)
val_size = int(0.15 * dataset_size)
```

```python
# Split Datasets
train_ds = raw_ds.take(train_size)
val_ds = raw_ds.skip(train_size).take(val_size)
test_ds = raw_ds.skip(train_size + val_size)

print(f"Total Batches: {dataset_size}")
print(f"Train Batches: {len(train_ds)}, Val Batches: {len(val_ds)}, Test Batches: {len(test_ds)}")
```

```python
# Updated Model with Dropout & Augmentation
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
])
```

```python
# 3. CNN Model
model = models.Sequential([
    layers.Rescaling(1./255, input_shape=(128, 128, 3)),
    
    # Data Augmentation apply karein
    data_augmentation,

    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    
    # Overfitting rokne ke liye Dropout
    layers.Dropout(0.5), 
    
    layers.Dense(2, activation='softmax')
])
```

```python

# 4. Model Compile
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)
```

```python
print("Training Shuru...")
history = model.fit(
    train_ds, 
    validation_data=val_ds, 
    epochs=15, 
    callbacks=[early_stopping]
)
```

```python
# 6. Testing (Ab yahan unseen Test Data par evaluate hoga)
print("\n--- Testing on Unseen Data ---")
test_loss, test_acc = model.evaluate(test_ds)
print(f"Final Test Accuracy: {test_acc * 100:.2f}%")
```

```python
# Naye recommended format (.keras) mein save karne ke liye:
model.save("mask_detector_model.keras")

print("Model successfully save ho gaya h!")
```
